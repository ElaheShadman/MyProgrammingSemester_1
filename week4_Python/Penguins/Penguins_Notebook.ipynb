 4. Cleaning the Data

 Purpose:
“Prepare clean and reliable data for analysis.”

    • Remove rows missing important numeric values
    • Replace missing “sex” values with “Unknown”

```python
df_clean = df.dropna(subset=["bill_length_mm", "flipper_length_mm"])
df_clean["sex"] = df_clean["sex"].fillna("Unknown")
```

![cleaning](week4_Python\Penguins\Cleaning.png)

5. Visualizing the Data

Purpose:
“Visualize patterns in the data: size differences, relationships, and habitat distribution.”

A) Histogram

Shows how bill length differs between species.

```python
sns.histplot(..., x="bill_length_mm", hue="species")
```

B) Boxplot

Shows flipper length distribution per species.

```python
sns.boxplot(..., x="species", y="flipper_length_mm")
```

C) Scatterplot

Shows relationship between bill length and body mass.

```python
sns.scatterplot(..., x="bill_length_mm", y="body_mass_g")
```

D) Countplot

Shows how many penguins of each species live on each island.

```python
sns.countplot(..., x="island", hue="species")
```

![Visualizing](week4_Python\Penguins\Visualizing.png)

6. Correlation Heatmap

Purpose:
“Identify which measurements are strongly related.”

    • Calculate correlations between numeric features
    • Plot them in a heatmap

```python
corr = df_clean[["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]].corr()
sns.heatmap(corr, annot=True)
```

![Heatmap](week4_Python\Penguins\Heatmap.png)
